```python
import pandas as pd

df = pd.read_csv(
    r"C:\Users\SHREE HARINI\Downloads\social_media_dataset_with_null_duplicates_outliers.csv"
)

print(df.head())
```

```python

import matplotlib.pyplot as plt


# Line chart
plt.figure(figsize=(10, 5))

plt.plot(df["Post_ID"], df["Shares"], marker="o")

plt.xlabel("Post ID")
plt.ylabel("Views")
plt.title("Views per Social Media Post")
plt.show()
```

```python
# Calculate average likes by platform
platform_likes = df.groupby("Platform")["Likes"].mean()

# Bar chart
plt.figure(figsize=(8, 5))

plt.bar(platform_likes.index, platform_likes.values)

plt.xlabel("Platform")
plt.ylabel("Average Likes")
plt.title("Average Likes by Platform")

plt.show()
```

```python
# Count sentiment categories
sentiment_count = df["Sentiment"].value_counts()

# Pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    sentiment_count.values,
    labels=sentiment_count.index,
    autopct="%1.1f%%"
)

plt.title("Sentiment Distribution of Social Media Posts")

plt.show()
```
